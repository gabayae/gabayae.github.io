# 1. Introduction to Time Series

A **time series** is a sequence of observations indexed in time order. This notebook introduces:
- Common types of time series: random walks, autoregressive processes, seasonal series
- Simulating time series in Python
- Understanding variance growth in non-stationary processes

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

%matplotlib inline
np.random.seed(42)
```

## 1.1 Random Walk

A **random walk** is defined as $X_t = X_{t-1} + \varepsilon_t$ where $\varepsilon_t \sim \mathcal{N}(0,1)$.
Equivalently, $X_t = \sum_{i=1}^t \varepsilon_i$, so $\text{Var}(X_t) = t\sigma^2$.
The variance grows linearly with time, making the process **non-stationary**.

```python
n = 500
eps = np.random.normal(0, 1, n)
rw = np.cumsum(eps)

plt.figure(figsize=(12, 4))
plt.plot(rw, lw=0.8)
plt.title('Random Walk')
plt.xlabel('t')
plt.grid(True, alpha=0.3)
plt.show()
```

## 1.2 AR(1) Process

The **AR(1)** process is $X_t = \phi X_{t-1} + \varepsilon_t$. When $|\phi| < 1$, the process is **stationary** with variance $\sigma^2/(1-\phi^2)$.

```python
phi = 0.8
ar1 = np.zeros(n)
for t in range(1, n):
    ar1[t] = phi * ar1[t-1] + eps[t]

plt.figure(figsize=(12, 4))
plt.plot(ar1, lw=0.8, color='red')
plt.title(f'AR(1) with $\\phi={phi}$')
plt.xlabel('t')
plt.grid(True, alpha=0.3)
plt.show()

print(f"Theoretical variance: {1/(1-phi**2):.4f}")
print(f"Empirical variance:   {np.var(ar1):.4f}")
```

## 1.3 Seasonal Series

Many real-world series exhibit **seasonality** (regular periodic patterns) and **trend** (long-term drift).
A classical decomposition is: $Y_t = T_t + S_t + \varepsilon_t$

```python
t = np.arange(1, 121)
trend = 0.05 * t
seasonal = 10 * np.sin(2 * np.pi * t / 12)
noise = np.random.normal(0, 2, 120)
y = 15 + trend + seasonal + noise

plt.figure(figsize=(12, 4))
plt.plot(t, y, lw=0.8, color='green')
plt.title('Seasonal Series (period=12) with Trend')
plt.xlabel('t')
plt.grid(True, alpha=0.3)
plt.show()
```

## 1.4 Comparing the Three Processes

```python
fig, axes = plt.subplots(3, 1, figsize=(12, 9))
axes[0].plot(rw, lw=0.8); axes[0].set_title('Random Walk (non-stationary)')
axes[1].plot(ar1, lw=0.8, color='red'); axes[1].set_title('AR(1) (stationary)')
axes[2].plot(t, y, lw=0.8, color='green'); axes[2].set_title('Seasonal + Trend')
plt.tight_layout()
plt.show()
```

## 1.5 Empirical Verification of Random Walk Variance

We verify theoretically that $\text{Var}(X_t) = t \cdot \sigma^2$ by simulating many realizations.

```python
rw_samples = np.cumsum(np.random.normal(0, 1, (1000, 200)), axis=1)
print(f"Var(X_200) theoretical: {200 * 1.0:.1f}")
print(f"Var(X_200) empirical:   {np.var(rw_samples[:, -1]):.1f}")
```

## Key Takeaways

- A **random walk** has growing variance and is non-stationary
- An **AR(1)** with $|\phi|<1$ is stationary with bounded variance
- Real-world series often combine **trend + seasonality + noise**
- **Stationarity** (constant mean and variance over time) is a key assumption for many models
