# 4. Exploratory Data Analysis (EDA)

EDA is the process of systematically examining a dataset to summarize its main characteristics, often using visualizations. This notebook demonstrates:
- Dataset overview and summary statistics
- Univariate analysis (distributions of individual variables)
- Bivariate analysis (relationships between variables)
- Correlation analysis
- Outlier detection with the IQR method

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

%matplotlib inline
```

## 4.1 Dataset Overview

We use the **Titanic** dataset. The first steps of any EDA are:
1. Check the shape and column types
2. Look at missing values
3. Compute basic summary statistics with `.describe()`

```python
titanic = sns.load_dataset('titanic')
print(f"Shape: {titanic.shape}")
print(f"\nColumn types:\n{titanic.dtypes}")
print(f"\nMissing values:\n{titanic.isnull().sum()}")
titanic.describe().round(2)
```

## 4.2 Univariate Analysis

Examine the distribution of the **target variable** (survived) and key features individually.

```python
print("Survival counts:")
print(titanic['survived'].value_counts())
print(f"\nSurvival rate: {titanic['survived'].mean():.2%}")

print("\nClass distribution:")
print(titanic['class'].value_counts())
```

## 4.3 Bivariate Analysis

Examine how features relate to the target. The **survival rate** varies dramatically across passenger class and sex.

```python
print("Survival by class:")
print(titanic.groupby('class')['survived'].mean().round(3))

print("\nSurvival by sex:")
print(titanic.groupby('sex')['survived'].mean().round(3))
```

## 4.4 EDA Visualizations

Visualizations complement numerical summaries. They reveal patterns that are hard to see in tables.

```python
fig, axes = plt.subplots(2, 2, figsize=(12, 10))

sns.countplot(data=titanic, x='class', hue='survived', ax=axes[0, 0])
axes[0, 0].set_title('Survival by Class')

sns.countplot(data=titanic, x='sex', hue='survived', ax=axes[0, 1])
axes[0, 1].set_title('Survival by Sex')

sns.histplot(data=titanic, x='age', hue='survived', kde=True,
             bins=30, ax=axes[1, 0])
axes[1, 0].set_title('Age Distribution by Survival')

sns.boxplot(data=titanic, x='class', y='fare', ax=axes[1, 1])
axes[1, 1].set_title('Fare by Class')

plt.tight_layout()
plt.show()
```

## 4.5 Correlation Heatmap

The correlation matrix quantifies linear relationships between all numerical features.

```python
numeric_cols = titanic.select_dtypes(include=[np.number])
corr = numeric_cols.corr()

plt.figure(figsize=(8, 6))
sns.heatmap(corr, annot=True, cmap='coolwarm', center=0, fmt='.2f')
plt.title('Titanic Numeric Correlations')
plt.tight_layout()
plt.show()
```

## 4.6 Outlier Detection (IQR Method)

The **Interquartile Range (IQR)** method defines outliers as values below $Q_1 - 1.5 \cdot IQR$ or above $Q_3 + 1.5 \cdot IQR$.

```python
Q1 = titanic['fare'].quantile(0.25)
Q3 = titanic['fare'].quantile(0.75)
IQR = Q3 - Q1
lower = Q1 - 1.5 * IQR
upper = Q3 + 1.5 * IQR

outliers = titanic[(titanic['fare'] < lower) | (titanic['fare'] > upper)]
print(f"Fare outliers (IQR method): {len(outliers)} out of {len(titanic)}")
print(f"IQR = {IQR:.2f}, bounds = [{lower:.2f}, {upper:.2f}]")
```

## Key Takeaways

1. Start EDA with `.shape`, `.dtypes`, `.describe()`, and `.isnull().sum()`
2. **Univariate**: understand each variable's distribution independently
3. **Bivariate**: use `groupby` and conditional plots to find relationships with the target
4. **Correlation heatmaps** reveal linear dependencies between numerical features
5. **IQR method** is a robust approach for flagging outliers
