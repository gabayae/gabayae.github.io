# 5. Your First Machine Learning Model

This notebook introduces the fundamental ML workflow using scikit-learn:
- Loading data and the train/test split
- Training a K-Nearest Neighbors (KNN) classifier
- Understanding the bias-variance tradeoff
- Cross-validation for robust evaluation

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.datasets import load_iris
from sklearn.model_selection import train_test_split, cross_val_score
from sklearn.neighbors import KNeighborsClassifier
from sklearn.metrics import accuracy_score

%matplotlib inline
```

## 5.1 Load and Explore the Data

The **Iris** dataset is a classic benchmark with 150 samples of 3 flower species, described by 4 numerical features.

```python
iris = load_iris(as_frame=True)
X = iris.data
y = iris.target
print(f"Features: {X.shape}")
print(f"Target classes: {iris.target_names}")
X.head()
```

## 5.2 Train/Test Split

We hold out 30% of data for testing. The `stratify` parameter ensures equal class proportions in both splits.

```python
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.3, random_state=42, stratify=y)
print(f"Train: {X_train.shape}, Test: {X_test.shape}")
```

## 5.3 KNN with Different k Values

The **K-Nearest Neighbors** algorithm classifies a point by majority vote among its $k$ nearest neighbors.
- **Small k** (e.g., 1): low bias, high variance (overfitting)
- **Large k** (e.g., 15): high bias, low variance (underfitting)

```python
k_values = [1, 3, 5, 7, 9, 11, 15]
train_scores = []
test_scores = []

for k in k_values:
    knn = KNeighborsClassifier(n_neighbors=k)
    knn.fit(X_train, y_train)
    tr_acc = knn.score(X_train, y_train)
    te_acc = knn.score(X_test, y_test)
    train_scores.append(tr_acc)
    test_scores.append(te_acc)
    print(f"  k={k:2d}: train={tr_acc:.3f}, test={te_acc:.3f}")
```

## 5.4 Bias-Variance Tradeoff Visualization

Plotting train vs test accuracy as a function of $k$ illustrates the classic tradeoff: the "sweet spot" balances underfitting and overfitting.

```python
fig, ax = plt.subplots(figsize=(8, 5))
ax.plot(k_values, train_scores, 'bo-', label='Train')
ax.plot(k_values, test_scores, 'rs-', label='Test')
ax.set_xlabel('k (number of neighbors)')
ax.set_ylabel('Accuracy')
ax.set_title('Bias-Variance Tradeoff (KNN)')
ax.legend()
ax.grid(True, alpha=0.3)
plt.show()
```

## 5.5 Cross-Validation

A single train/test split can be misleading. **K-fold cross-validation** provides a more reliable estimate by averaging over multiple splits.

```python
knn5 = KNeighborsClassifier(n_neighbors=5)
cv_scores = cross_val_score(knn5, X, y, cv=5, scoring='accuracy')

print(f"5-Fold CV Scores: {cv_scores.round(3)}")
print(f"Mean: {cv_scores.mean():.3f} (+/- {cv_scores.std():.3f})")
```

## Key Takeaways

1. The ML workflow: **load data -> split -> train -> evaluate**
2. KNN is a simple but effective baseline classifier
3. The **bias-variance tradeoff** is fundamental: tune hyperparameters (like $k$) to balance it
4. **Cross-validation** gives a more reliable performance estimate than a single split
5. Always use `stratify` when splitting classification data
