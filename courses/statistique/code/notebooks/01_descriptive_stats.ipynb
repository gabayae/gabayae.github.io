# 1. Descriptive Statistics

Descriptive statistics summarize and describe the main features of a dataset. This notebook covers:
- Measures of central tendency: mean, median, mode
- Measures of dispersion: variance, standard deviation, IQR, coefficient of variation
- Shape measures: skewness and kurtosis
- Visualizations: histogram, boxplot, ECDF
- Bivariate correlation

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from scipy import stats

%matplotlib inline
```

## 1.1 Sample Data

We work with a small salary dataset to illustrate all concepts clearly.

```python
salaires = np.array([1800, 1950, 2100, 2100, 2200, 2300, 2350, 2400,
                     2500, 2600, 2650, 2700, 2800, 2900, 3100, 3200,
                     3500, 3800, 4500, 8000])
print(f"n = {len(salaires)}")
print(f"Range: [{salaires.min()}, {salaires.max()}]")
```

## 1.2 Measures of Central Tendency

- **Mean** ($\bar{x}$): arithmetic average, sensitive to outliers
- **Median**: middle value, robust to outliers
- **Mode**: most frequent value

When the data is right-skewed: $\text{mode} < \text{median} < \text{mean}$

```python
print(f"Mean   : {np.mean(salaires):.2f}")
print(f"Median : {np.median(salaires):.2f}")
print(f"Mode   : {stats.mode(salaires, keepdims=True).mode[0]}")
```

## 1.3 Measures of Dispersion

- **Variance** ($s^2$): average squared deviation from the mean
- **Standard deviation** ($s$): square root of variance (same units as data)
- **IQR** = $Q_3 - Q_1$: robust to outliers
- **CV** = $s/\bar{x}$: dimensionless measure for comparing variability

```python
print(f"Std (ddof=1): {np.std(salaires, ddof=1):.2f}")
print(f"Variance    : {np.var(salaires, ddof=1):.2f}")
print(f"Range       : {np.ptp(salaires)}")
print(f"IQR         : {stats.iqr(salaires):.2f}")
print(f"CV          : {np.std(salaires, ddof=1)/np.mean(salaires):.4f}")
```

## 1.4 Shape: Skewness and Kurtosis

- **Skewness** > 0: right-skewed (tail on the right)
- **Excess kurtosis** > 0: heavier tails than a normal distribution

```python
print(f"Skewness        : {stats.skew(salaires):.4f}")
print(f"Kurtosis (excess): {stats.kurtosis(salaires):.4f}")
```

## 1.5 Visualizations

```python
fig, axes = plt.subplots(1, 3, figsize=(15, 4))

# Histogram
axes[0].hist(salaires, bins=8, edgecolor='black', alpha=0.7, color='steelblue')
axes[0].axvline(np.mean(salaires), color='red', ls='--', label=f'Mean={np.mean(salaires):.0f}')
axes[0].axvline(np.median(salaires), color='green', ls='-', label=f'Median={np.median(salaires):.0f}')
axes[0].set_title('Histogram')
axes[0].legend(fontsize=8)

# Boxplot
bp = axes[1].boxplot(salaires, vert=True, patch_artist=True)
bp['boxes'][0].set_facecolor('lightblue')
axes[1].set_title('Boxplot')

# ECDF
x_sorted = np.sort(salaires)
y_ecdf = np.arange(1, len(x_sorted)+1) / len(x_sorted)
axes[2].step(x_sorted, y_ecdf, where='post', color='navy', lw=2)
axes[2].set_title('Empirical CDF')
axes[2].set_ylabel('$F_n(x)$')
axes[2].grid(True, alpha=0.3)

plt.tight_layout()
plt.show()
```

## 1.6 Bivariate Analysis: Correlation

The **Pearson correlation coefficient** $r$ measures linear association between two variables. $r \in [-1, 1]$.

```python
x = np.array([8, 12, 14, 10, 16, 6, 18, 11])
y = np.array([9, 13, 15, 12, 17, 8, 19, 13])

r, p_val = stats.pearsonr(x, y)
print(f"Pearson r = {r:.4f}, p-value = {p_val:.6f}")
```

## Key Takeaways

- The **mean** is sensitive to outliers; the **median** is robust
- **Standard deviation** and **IQR** measure spread; **CV** allows comparison across scales
- **Skewness** and **kurtosis** describe the shape of the distribution
- Always combine numerical summaries with **visualizations** (histogram, boxplot, ECDF)
